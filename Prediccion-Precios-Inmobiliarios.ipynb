<h1>Ejercicio 1:</h1>
<h3>¿Cuál es la diferencia entre el aprendizaje supervisado y el aprendizaje no
supervisado?</h3>

<p> <b>Respuesta:</b> En el aprendizaje supervisado se etiquetan los datos; el algorimo clasifica los
datos de acuerdo a grupos supeditados. En el aprendizaje no supervisado los datos no vienen etiquetados;
el algoritmo agrupa los datos en clusters o "grupos" de acuerdo a características comunes que luego un
humano puede clasificar según estime conveniente para su proyecto.</p>

<h1>Ejercicio 3:</h1>
<h3> ¿Por qué razón es necesario dividir el set de datos en 2 set de datos al momento de
realizar algún modelamiento supervisado? ¿Cuáles son los 2 set de datos y cúal es la
función de cada uno?</h3>

<p> <b>Respuesta: </b> Es necesario separar los set de datos porque cada uno cumple una función distintas. Un conjunto de datos debe utilizarse para entrenar al modelo y otro <em>distinto</em> debe utilizarse para evaluar al modelo. El modelo para entrenar sirve para que el algoritmo pueda hacer distinciones entre los datos, pero si se evalúa al modelo con los mismos datos de entrenamiento, no se puede comprobar si realmente aprendió o si sólo "memorizó" los datos. Así, la función del conjunto de evaluación consiste en verificar que el algoritmo puede, efectivamente, clasificar datos distintos a los de entrenamiento y no sólo replicar lo que ya conoce. </p>

<h1>Ejercicio 8:</h1>

<h3> Genera un modelo de regresión lineal. Para ello: <br>
a. Importa el estimador de regresión lineal de Scikit-learn. <br>
b. Divide los datos (80% train y 20% test) <br>
c. Crea unainstancia del modelo de regresión lineal. <br>
d. Ajusta el modelo utilizando los datos de entrenamiento.</h3>

```python
# Vamos a tomar la variable Area como Feature y la variable Price como Target.

x = df[["area"]] # Definimos el feature como un Dataframe con doble corchete
y = df["price"]  # Definimos el target como una serie o lista con corchete simple

# a) Importamos la función de la librería para realizar el entrenamiento y testeo
from sklearn.model_selection import train_test_split
from sklearn import linear_model

# b) Separamos datos de entrenamiento y de evaluación en razón 80% - 20% 
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=123)

# c) Instanciamos el modelo

regresion = linear_model.LinearRegression() # Se crea una instancia de regresión lineal con los datos que preparamos

# d) Ajustar el modelo

regresion.fit(X_train, y_train) # Se crea la ecuación de la recta de regresión
```

<h1>Ejercicio 9:</h1>
<h3> Evalúa el modelo. Para ello: <br>
a. Utiliza los datos de entrenamiento y calcula métricas de evaluación, como el
error cuadrado medio (MSE) o el coeficiente de determinación (R^2). Grafica
los coeficientes. <br>
b. Utiliza los datos de prueba para realizar predicciones con el modelo
entrenado, y evalúa el rendimiento del modelo comparando las predicciones
con los valores reales</h3>

```python
from sklearn.metrics import mean_squared_error, r2_score # Importamos métricas de R.L.

# a) Realizamos una predicción y calculamos y mostramos métricas
y_pred = regresion.predict(X_test)
print("Coeficientes: \n", regresion.coef_)
print('\n Error cuadrado medio: %.2f' % mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
print("Coeficiente de determinación (R^2):", r2)

# b) Mostramos un gráfico con predicciones y valores reales
plt.scatter(y_test, y_pred)
plt.xlabel("Valores reales")
plt.ylabel("Predicciones")
plt.title("Predicciones vs Valores reales")
plt.show()
```

```python
# En el ejercicio anterior, podemos observar un R2 de 0.19... eso significa que el modelo
# sólo explica el 19% del precio... Si, por ejemplo, probáramos con una regresión multilineal
# podríamos obtener una mayor explicación del modelo.

x = df[["area", "bathrooms", "bedrooms"]] # Definimos el feature con múltiples variables, pero no todas, dejando margen de mejora.
y = df["price"]  # Definimos el target como antes

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=123) # Dividimos datos de entrenamiento y test
regresion = linear_model.LinearRegression()  # Creamos un nuevo modelo de regresión, que ahora es multilineal
regresion.fit(X_train, y_train) # ajustamos datos de entrenamiento
# Finalmente, repetimos los pasos del ejercicio 9:

y_pred = regresion.predict(X_test)
print("Coeficientes: \n", regresion.coef_)
print('\n Error cuadrado medio: %.2f' % mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
print("Coeficiente de determinación (R^2):", r2)
plt.scatter(y_test, y_pred)
plt.xlabel("Valores reales")
plt.ylabel("Predicciones")
plt.title("Predicciones vs Valores reales")
plt.show()

# Ahora, nuestro modelo explica casi el 50% del precio. La mejora es evidente. El margen de mejora también.
```
